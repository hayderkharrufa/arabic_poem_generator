```python
import rhymes
import markovify
from tqdm import tqdm
```

```python
def markov(text_file):
    with open(text_file, 'r', encoding='utf-8') as f:
        text = f.read()
    text_model = markovify.NewlineText(text)
    return text_model
```

```python
def sample_bayts(text_model, iterations, use_tqdm):
    for _ in tqdm(range(iterations)) if use_tqdm else range(iterations):
        bayt = text_model.make_short_sentence(280, tries=100)
        if bayt:
            yield bayt, bayt.split()[-1]

def generate_poem_single_rhyme(poet_name, rhyme, iterations=3000, use_tqdm=False):
    text_model = markov('input/{}.txt'.format(poet_name))
    rhymes_list = set(rhymes.rhymes_with_last_n_chars(rhyme, len(rhyme)))
    bayts = {}

    for bayt, last_word in sample_bayts(text_model, iterations, use_tqdm):
        if last_word in rhymes_list and last_word not in bayts:
            bayts[last_word] = bayt
            if not use_tqdm:
                print(bayt)
    return '\n'.join(bayts.values())

def generate_poem_2_rhymes(poet_name, rhyme_1, rhyme_2, iterations=3000, use_tqdm=False):
    text_model = markov('input/{}.txt'.format(poet_name))
    rhymes_1_list = set(rhymes.rhymes_with_last_n_chars(rhyme_1, len(rhyme_1)))
    rhymes_2_list = set(rhymes.rhymes_with_last_n_chars(rhyme_2, len(rhyme_2)))
    bayts_1 = {}
    bayts_2 = {}

    for bayt, last_word in sample_bayts(text_model, iterations, use_tqdm):
        if last_word in rhymes_1_list:
            bayts_1.setdefault(last_word, bayt)
        if last_word in rhymes_2_list:
            bayts_2.setdefault(last_word, bayt)

    return '\n'.join('{}\n{}'.format(b1, b2) for b1, b2 in zip(bayts_1.values(), bayts_2.values()))
```
